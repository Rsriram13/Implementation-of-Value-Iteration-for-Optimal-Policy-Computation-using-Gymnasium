```python
import gymnasium as gym
import numpy as np
import matplotlib.pyplot as plt
```

```python
# -------------------------------------------------
# Create FrozenLake Environment
# -------------------------------------------------
env_desc = [
    "FSHF",
    "FHFH",
    "HFFH",
    "HHFG"
]

env = gym.make("FrozenLake-v1", desc=env_desc, is_slippery=True)
# write your code here
env = env.unwrapped
```

```python
# -------------------------------------------------
# Value Iteration Algorithm (Combined Function)
# -------------------------------------------------

def value_iteration(env, gamma=0.99, theta=1e-8):
    """
    Performs value iteration and extracts the optimal policy simultaneously.
    """
    num_states = env.observation_space.n
    num_actions = env.action_space.n

    V = np.zeros(num_states)
    policy = np.zeros(num_states, dtype=int)
    iterations = 0

    while True:
        delta = 0
        V_new = np.copy(V)
        for s in range(num_states):
            q_values = np.zeros(num_actions)
            for a in range(num_actions):
                for prob, next_state, reward, done in env.P[s][a]:
                    q_values[a] += prob * (reward + gamma * V[next_state])

            V_new[s] = np.max(q_values)
            policy[s] = np.argmax(q_values)  # Derives policy action directly during evaluation
            delta = max(delta, np.abs(V_new[s] - V[s]))

        V = V_new
        iterations += 1

        if delta < theta:
            break

    return V, policy, iterations
```

```python
# -------------------------------------------------
# Run Value Iteration
# -------------------------------------------------

# write your code here
V, policy, iterations = value_iteration(env)
```

```python

# -------------------------------------------------
# Display Output
# -------------------------------------------------
print("Name: Sriram R")
print("Register Number:212223230215")
print("Value Iteration Completed")
print("Number of Iterations:", iterations)

print("\nOptimal State-Value Function:")
print(np.round(V.reshape(4, 4), 4))


action_symbols = {
    0: "L",
    1: "D",
    2: "R",
    3: "U"
}

policy_grid = np.array(
    [action_symbols[action] for action in policy]
).reshape(4, 4)

print("\nOptimal Policy:")
print(policy_grid)

env.close()


```
